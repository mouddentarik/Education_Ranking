```python
import pandas as pd

# Read the Excel file without using caas_jupyter_tools
excel_path = "olympiad_table.xlsx"
df_olymp = pd.read_excel(excel_path)

# Simply show the first few rows as output
df_olymp.head(10)

```

```python
# Display column names and their data types
df_olymp.dtypes
```

```python
df_olymp.shape
```

```python
# Remove all rows that contain at least one missing value in any column
df_olymp_clean = df_olymp.dropna()

# Display the shape and first few rows to confirm
df_olymp_clean.shape
```

```python
df_olymp_clean.head()
```

```python
import matplotlib.pyplot as plt

# Select only the columns 25 to 20 (in reverse chronological order as provided)
cols_to_plot = ['25', '24', '23', '22', '21', '20']

# Plot each country's line on the same chart
plt.figure(figsize=(10, 6))
for _, row in df_olymp_clean.iterrows():
    plt.plot(cols_to_plot, row[cols_to_plot], marker='o', label=row['Country'])

plt.title("Country Performance (Years 2020–2025 Equivalent)")
plt.xlabel("Year (last two digits)")
plt.ylabel("Value / Rank")
plt.legend(title="Country")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Filter the desired countries
countries_to_plot = ['MAR', 'SAF', 'ALG', 'URY', 'SWE']
df_selected = df_olymp[df_olymp['Country'].isin(countries_to_plot)]

# Define columns for years 25 to 20
cols_to_plot = ['25', '24', '23', '22', '21', '20']

# Plot lines for each selected country
plt.figure(figsize=(10, 6))
for _, row in df_selected.iterrows():
    plt.plot(cols_to_plot, row[cols_to_plot], marker='o', label=row['Country'])

plt.title("Performance of Selected Countries (Years 2020–2025 Equivalent)")
plt.xlabel("Year (last two digits)")
plt.ylabel("Place")
plt.legend(title="Country")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Filter the desired countries
countries_to_plot = ['MAR', 'SAF', 'ALG', 'URY', 'SWE']
df_selected = df_olymp[df_olymp['Country'].isin(countries_to_plot)]

# Define columns for years 25 to 20
cols_to_plot = ['25', '24', '23', '22', '21', '20']

# Plot lines for each selected country with annotations, skipping NaNs
plt.figure(figsize=(10, 6))
for _, row in df_selected.iterrows():
    plt.plot(cols_to_plot, row[cols_to_plot], marker='o', label=row['Country'])
    for x, y in zip(cols_to_plot, row[cols_to_plot]):
        if not np.isnan(y):
            plt.text(x, y, f'{int(y)}', fontsize=9, ha='left', va='bottom')

plt.title("Performance of Selected Countries (Years 2020–2025 Equivalent)")
plt.xlabel("Year (last two digits)")
plt.ylabel("Place")
plt.legend(title="Country")
plt.grid(True)
plt.show()
```
